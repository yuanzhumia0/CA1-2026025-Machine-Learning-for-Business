# Introduction

This project focuses on the **transport domain** using the 2025 NYC Yellow Taxi dataset **(New York City Taxi and Limousine Commission, 2025)**.

Taxi trip data contains useful information about travel distance, trip duration, speed, cost and daily demand. Analysing these patterns can help identify different types of taxi journeys and understand how taxi demand changes over time.

The dataset was selected because it is a large real-world transport dataset with millions of observations and multiple numerical and categorical variables. This makes it suitable for both **clustering analysis** and **time series forecasting**.

The main objectives of this project are:

1. To clean and prepare the NYC Yellow Taxi dataset for machine learning analysis.
2. To use clustering algorithms to identify different groups of taxi trips based on trip characteristics.
3. To compare **K-Means** and **Agglomerative Hierarchical Clustering** using the Silhouette Score and Davies-Bouldin Index.
4. To explore daily taxi demand for trends, seasonality and unusual observations.
5. To build and evaluate a SARIMA model for forecasting future daily taxi trip demand.
6. To interpret the results and identify the strengths and limitations of the selected models.

Overall, the project aims to use unsupervised machine learning and time series forecasting to better understand taxi trip behaviour and daily transport demand.

### Time-Based Data Split

The dataset was split based on time to avoid data leakage.

- **January to June:** Training set
- **July to September:** Validation set
- **October to December:** Test set

The training set is used to train the model. The validation set is used to compare models and tune parameters. The test set is used at the end to evaluate the final model on unseen future data.

# Part 1: Clustering Analysis

In this section, clustering algorithms will be used to identify different groups of NYC Yellow Taxi trips based on their trip characteristics.

```python
# Select features for clustering
clustering_features = [
    "trip_distance",
    "trip_duration_minutes",
    "distance_per_hour",
    "total_amount"
]

df_clustering = df_analysis[clustering_features].copy()

print("Clustering data shape:", df_clustering.shape)
df_clustering.head()
```

### Clustering Algorithm Selection

Two clustering algorithms were selected for this project: **K-Means** and **Agglomerative Hierarchical Clustering**.

K-Means was selected because it is efficient and suitable for large numerical datasets (Scikit-learn Developers, 2026). The NYC Yellow Taxi dataset contains millions of records, so K-Means is useful for grouping taxi trips based on similarities in distance, duration, speed and cost.

**Agglomerative Hierarchical Clustering** was selected as a second method because it creates clusters using a hierarchical approach **(Scikit-learn Developers, 2026)**. This provides a different way of grouping the taxi trips and allows the results to be compared with K-Means.

Because Agglomerative Clustering requires more memory and processing time, a smaller sample was used for this algorithm.

The performance of both methods will be compared using the **Silhouette Score** and **Davies-Bouldin Index**.

### Clustering Features

Four numerical features were selected for the clustering analysis:

- `trip_distance`
- `trip_duration_minutes`
- `distance_per_hour`
- `total_amount`

These features describe the distance, duration, speed and cost of each taxi trip.

Clustering is an unsupervised learning method, so there is no target variable.

```python
# Sample data for clustering
df_clustering_sample = df_clustering.sample(
    n=100000,
    random_state=42
)

print("Clustering sample shape:", df_clustering_sample.shape)

df_clustering_sample.head()
```

### Clustering Sample

The cleaned dataset contains more than 21 million taxi trips.

A random sample of **100,000 records** was selected for the clustering analysis to reduce processing time and memory usage.

The sample contains the same four clustering features:

- `trip_distance`
- `trip_duration_minutes`
- `distance_per_hour`
- `total_amount`

The sampling used `random_state=42` so that the same records can be selected again when the code is rerun.

```python
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Standardize the clustering sample
clustering_scaled = scaler.fit_transform(df_clustering_sample)


# Convert scaled data back to a DataFrame
df_clustering_scaled = pd.DataFrame(
    clustering_scaled,
    columns=clustering_features
)

print("Scaled data shape:", clustering_scaled.shape)
df_clustering_scaled.head()
```

### Feature Scaling

The four clustering features have different numerical ranges.

`StandardScaler` was used to standardize the features before applying the clustering algorithms **(Scikit-learn Developers, 2026)**.

This gives each feature a similar scale, so variables with larger values do not have too much influence on the clustering results.

After scaling, the dataset contains **100,000 records and 4 standardized features**.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Test different numbers of clusters
k_values = range(2, 7)
inertia_values = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(df_clustering_scaled)
    inertia_values.append(kmeans.inertia_)

print("K values:", list(k_values))
print("Inertia values:", inertia_values)
```

```python
plt.figure(figsize=(7, 5))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for K-Means")

plt.show()
```

### Elbow Method Results

The Elbow Method was used to test K-Means with different numbers of clusters from **2 to 6**.

The inertia decreased as the number of clusters increased.

The largest reduction happened between **k = 2** and **k = 3**, and the rate of improvement became smaller after this point.

Based on the elbow graph, **k = 3** appears to be a reasonable number of clusters.

However, the Silhouette Score and Davies-Bouldin Index will also be used to evaluate the clustering performance before selecting the final value of `k`.

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score

silhouette_scores = []
db_scores = []

for k in range(2, 7):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(df_clustering_scaled)

    # Silhouette Score
    sil_score = silhouette_score(
        df_clustering_scaled,
        labels,
        sample_size=10000,
        random_state=42
    )

    # Davies-Bouldin Index
    db_score = davies_bouldin_score(
        df_clustering_scaled,
        labels
    )

    silhouette_scores.append(sil_score)
    db_scores.append(db_score)

    print(
        f"k={k} | "
        f"Silhouette Score={sil_score:.4f} | "
        f"Davies-Bouldin Index={db_score:.4f}"
    )
```

### K-Means Evaluation Results

The K-Means models were evaluated using the Silhouette Score and Davies-Bouldin Index for different values of `k`.

The results show that **k = 2** achieved the best clustering performance.

- **Silhouette Score:** 0.6898
- **Davies-Bouldin Index:** 0.7095

The Silhouette Score was highest for `k = 2`, while the Davies-Bouldin Index was also the lowest.

Therefore, **2 clusters were selected for the final K-Means model**.

```python
# Train the final K-Means model
final_kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)  

cluster_labels = final_kmeans.fit_predict(df_clustering_scaled)

df_clustering_sample["Cluster"] = cluster_labels

print(df_clustering_sample["Cluster"].value_counts())
```

### Final K-Means Clustering

The final K-Means model was trained using **2 clusters**, based on the previous clustering evaluation results.

The 100,000 sampled taxi trips were divided into:

- **Cluster 0:** 87,750 trips
- **Cluster 1:** 12,250 trips

Cluster 0 contains most of the taxi trips, while Cluster 1 represents a smaller group.

Further analysis is needed to understand the characteristics of each cluster.

```python
# Compare the characteristics of each cluster
cluster_summary = df_clustering_sample.groupby("Cluster")[
    [
        "trip_distance",
        "trip_duration_minutes",
        "distance_per_hour",
        "total_amount"
    ]
].mean()

cluster_summary
```

### Silhouette Score and Davies-Bouldin Index

Two evaluation metrics were used to measure clustering performance: the **Silhouette Score** and the **Davies-Bouldin Index**.

The **Silhouette Score** measures how similar each observation is to its own cluster compared with other clusters **(Scikit-learn Developers, 2026)**.  The score ranges approximately from **-1 to 1**. A higher value indicates that observations are well grouped within their own cluster and well separated from other clusters.

The **Davies-Bouldin Index** compares the similarity between clusters by considering the compactness within each cluster and the separation between different clusters **(Scikit-learn Developers, 2026)**.  A lower value indicates better clustering performance.

Therefore:

- **Higher Silhouette Score = better**
- **Lower Davies-Bouldin Index = better**

Using both metrics provides a more complete evaluation because they assess clustering quality from different perspectives.

```python
# Calculate anomaly thresholds using the IQR method
Q1 = daily_trips.quantile(0.25)
Q3 = daily_trips.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find unusual daily trip counts
anomalies = daily_trips[
    (daily_trips < lower_bound) |
    (daily_trips > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of anomalies:", len(anomalies))

anomalies
```

### Anomaly Analysis

```python
from statsmodels.tsa.stattools import adfuller

# Augmented Dickey-Fuller test using training data only
adf_result = adfuller(
    train_daily,
    result_object=False
)

print("ADF Statistic:", adf_result[0])
print("p-value:", adf_result[1])
print("Critical Values:")

for key, value in adf_result[4].items():
    print(f"{key}: {value}")
```

```python
# 第-4 First difference of the training time series
train_daily_diff = train_daily.diff().dropna()

# ADF test after first differencing
adf_diff_result = adfuller(
    train_daily_diff,
    result_object=False
)

print("ADF Statistic:", adf_diff_result[0])
print("p-value:", adf_diff_result[1])
print("Critical Values:")

for key, value in adf_diff_result[4].items():
    print(f"{key}: {value}")
```

### Stationarity Test

The Augmented Dickey-Fuller (ADF) test was applied to the **January to June training data** to avoid using future validation or test information **(Statsmodels Developers, 2026)**.

For the original training time series:

- **ADF Statistic:** -2.7202
- **p-value:** 0.0706

Because the p-value was greater than **0.05**, the original training series was not considered stationary at the 5% significance level.

A first difference was then applied to the training data. After differencing:

- **ADF Statistic:** -4.6670
- **p-value:** 0.000097

The p-value is now below **0.05**, indicating that the differenced series is stationary.

Therefore, one level of differencing is appropriate and **d = 1** was selected for the SARIMA model.

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

# 第-5 ACF plot using differenced training data
plot_acf(
    train_daily_diff,
    lags=30
)

plt.title("ACF of Differenced Training Taxi Trips")
plt.show()

# PACF plot using differenced training data
plot_pacf(
    train_daily_diff,
    lags=30,
    method="ywm"
)

plt.title("PACF of Differenced Training Taxi Trips")
plt.show()
```

### ACF and PACF Analysis

The ACF and PACF plots were examined using the **first-differenced January to June training data** **(Statsmodels Developers, 2026)**.

The ACF plot shows strong positive correlations at approximately **lags 7, 14, 21 and 28**. This indicates a clear repeating weekly pattern in the taxi demand data.

The PACF plot also contains significant correlations at several lags, including a noticeable seasonal relationship around **lag 7**.

The ADF test showed that one level of differencing was required, so **d = 1** was selected.

The strong correlations at multiples of 7 support the use of a **7-day seasonal period** in the SARIMA model.

Because the ACF and PACF plots do not indicate one single clear combination of `p` and `q`, several small SARIMA parameter combinations will be compared using the validation dataset before selecting the final model.

```python
# Split the daily time series into training, validation and test sets

# January to June
ts_train = daily_trips[
    (daily_trips.index >= "2025-01-01") &
    (daily_trips.index < "2025-07-01")
]

# July to September
ts_validation = daily_trips[
    (daily_trips.index >= "2025-07-01") &
    (daily_trips.index < "2025-10-01")
]

# October to December
ts_test = daily_trips[
    daily_trips.index >= "2025-10-01"
]

print("Training days:", len(ts_train))
print("Validation days:", len(ts_validation))
print("Test days:", len(ts_test))

print("\nTraining period:")
print(ts_train.index.min(), "to", ts_train.index.max())

print("\nValidation period:")
print(ts_validation.index.min(), "to", ts_validation.index.max())

print("\nTest period:")
print(ts_test.index.min(), "to", ts_test.index.max())
```

### Time Series Train-Validation-Test Split

The daily taxi trip time series was divided chronologically into training, validation and test datasets.

- **Training period:** January to June 2025 — **181 days**
- **Validation period:** July to September 2025 — **92 days**
- **Test period:** October to December 2025 — **92 days**

The training data is used to fit the SARIMA models, the validation data is used to compare and select the best model, and the test data is reserved for the final evaluation.

The data was not randomly shuffled because the chronological order must be preserved for time series forecasting.

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Set an explicit daily frequency
ts_train = ts_train.asfreq("D")
ts_validation = ts_validation.asfreq("D")
ts_test = ts_test.asfreq("D")

# Create the initial SARIMA model
sarima_model = SARIMAX(
    ts_train,
    order=(1, 1, 1),
    seasonal_order=(1, 0, 1, 7)
)

# Train the model using January to June data
sarima_result = sarima_model.fit(
    disp=False,
    maxiter=500
)

print("SARIMA model training completed")
print("AIC:", sarima_result.aic)
print("Converged:", sarima_result.mle_retvals["converged"])
```

### SARIMA Model Training

An initial SARIMA model was trained using the January to June 2025 training data **(Statsmodels Developers, 2026)**.

Based on the stationarity test, one level of differencing was required, so the initial model used:

- **Non-seasonal order:** (1, 1, 1)
- **Seasonal order:** (1, 0, 1, 7)

The seasonal period of **7 days** was selected because the earlier analysis showed a clear weekly seasonal pattern.

The model achieved an **AIC of 3795.42** and successfully converged.

This initial model was then evaluated using the July to September validation dataset before comparing several SARIMA parameter combinations.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Forecast the validation period
sarima_validation_forecast = sarima_result.get_forecast(
    steps=len(ts_validation)
)

y_validation_forecast = sarima_validation_forecast.predicted_mean

# Match forecast dates with the validation dataset
y_validation_forecast.index = ts_validation.index

# Evaluate the validation forecast
sarima_validation_mae = mean_absolute_error(
    ts_validation,
    y_validation_forecast
)

sarima_validation_rmse = np.sqrt(
    mean_squared_error(
        ts_validation,
        y_validation_forecast
    )
)

print("Validation MAE:", sarima_validation_mae)
print("Validation RMSE:", sarima_validation_rmse)
```

```python
# Compare actual and forecast values on validation data
validation_comparison = pd.DataFrame({
    "Actual": ts_validation,
    "Forecast": y_validation_forecast
})

validation_comparison["Difference"] = abs(
    validation_comparison["Actual"] -
    validation_comparison["Forecast"]
)

validation_comparison.head(10)
```

### SARIMA Validation Forecast Evaluation

The initial **SARIMA (1,1,1) × (1,0,1,7)** model was evaluated using the July to September 2025 validation dataset.

The model achieved:

- **MAE:** 35,105.13 trips
- **RMSE:** 38,394.23 trips

The MAE means that the forecast differed from the actual daily taxi trip counts by about **35,105 trips on average**.

The RMSE is higher than the MAE, which indicates that some days had larger forecasting errors.

These results showed that the initial model was not sufficiently accurate, so additional SARIMA parameter combinations were compared using the same validation dataset.

```python
plt.figure(figsize=(14, 6))

plt.plot(
    ts_validation.index,
    ts_validation.values,
    label="Actual"
)

plt.plot(
    y_validation_forecast.index,
    y_validation_forecast.values,
    label="Forecast"
)

plt.xlabel("Date")
plt.ylabel("Number of Trips")
plt.title("SARIMA Validation Forecast vs Actual Taxi Trips")

plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

### Validation Forecast Strengths and Limitations

The initial **SARIMA (1,1,1) × (1,0,1,7)** model was able to capture some of the repeating weekly pattern in taxi demand.

However, the validation forecast was much smoother than the actual daily trip counts and did not capture several larger increases and decreases in demand.

The validation results were:

- **MAE:** 35,105.13 trips
- **RMSE:** 38,394.23 trips

The relatively high MAE and RMSE showed that the initial SARIMA model was not accurate enough for the final forecast.

Therefore, additional SARIMA parameter combinations were tested using the validation dataset to find a better-performing model.

```python
# Compare different SARIMA models using the validation dataset

sarima_candidates = [
    ((1, 1, 0), (1, 0, 1, 7)),
    ((0, 1, 1), (1, 0, 1, 7)),
    ((1, 1, 1), (1, 0, 1, 7)),
    ((1, 1, 2), (1, 0, 1, 7))
]

sarima_results = []

for order, seasonal_order in sarima_candidates:

    model = SARIMAX(
        ts_train,
        order=order,
        seasonal_order=seasonal_order
    )

    result = model.fit(
        disp=False,
        maxiter=500
    )

    forecast = result.get_forecast(
        steps=len(ts_validation)
    ).predicted_mean

    forecast.index = ts_validation.index

    mae = mean_absolute_error(
        ts_validation,
        forecast
    )

    rmse = np.sqrt(
        mean_squared_error(
            ts_validation,
            forecast
        )
    )

    converged = result.mle_retvals["converged"]

    sarima_results.append([
        order,
        seasonal_order,
        result.aic,
        mae,
        rmse,
        converged
    ])

sarima_comparison = pd.DataFrame(
    sarima_results,
    columns=[
        "Order",
        "Seasonal Order",
        "AIC",
        "Validation MAE",
        "Validation RMSE",
        "Converged"
    ]
)

sarima_comparison
```

### Final SARIMA Model Selection

Several SARIMA models with **d = 1** were compared using the July to September validation dataset.

The best-performing model was:

**SARIMA (1,1,0) × (1,0,1,7)**

Its validation results were:

- **AIC:** 3784.38
- **Validation MAE:** 13,094.29 trips
- **Validation RMSE:** 15,733.34 trips
- **Converged:** True

This model achieved the lowest validation MAE and RMSE among the tested models and also had the lowest AIC.

Therefore, **SARIMA (1,1,0) × (1,0,1,7)** was selected as the final forecasting model.

```python
# 第-6 Combine training and validation data for the final model
ts_final_train = pd.concat([
    ts_train,
    ts_validation
]).sort_index()

ts_final_train = ts_final_train.asfreq("D")

# Train the selected final SARIMA model
final_sarima_model = SARIMAX(
    ts_final_train,
    order=(1, 1, 0),
    seasonal_order=(1, 0, 1, 7)
)

final_sarima_result = final_sarima_model.fit(
    disp=False,
    maxiter=500
)

print("Final training days:", len(ts_final_train))
print("Final training period:")
print(ts_final_train.index.min(), "to", ts_final_train.index.max())

print("\nFinal SARIMA AIC:", final_sarima_result.aic)
print(
    "Converged:",
    final_sarima_result.mle_retvals["converged"]
)
```

```python
# Forecast the final test period
final_test_forecast = final_sarima_result.get_forecast(
    steps=len(ts_test)
)

y_test_forecast = final_test_forecast.predicted_mean

# Match forecast dates with the test dataset
y_test_forecast.index = ts_test.index

# Evaluate final test performance
final_test_mae = mean_absolute_error(
    ts_test,
    y_test_forecast
)

final_test_rmse = np.sqrt(
    mean_squared_error(
        ts_test,
        y_test_forecast
    )
)

print("Final Test MAE:", final_test_mae)
print("Final Test RMSE:", final_test_rmse)
```

### Final Test Evaluation

The selected **SARIMA (1,1,0) × (1,0,1,7)** model was retrained using the January to September data and then evaluated on the unseen October to December test dataset.

The final test results were:

- **MAE:** 17,972.73 trips
- **RMSE:** 26,354.91 trips

The MAE indicates that the forecast differed from the actual daily taxi demand by approximately **17,973 trips per day on average**.

The RMSE is higher than the MAE, which suggests that some days had relatively large forecasting errors.

Overall, the model captured the regular time series structure, but its forecasting accuracy decreased when larger or unusual changes in taxi demand occurred.

```python
# Plot final test actual vs forecast values

plt.figure(figsize=(14, 6))

plt.plot(
    ts_test.index,
    ts_test.values,
    label="Actual"
)

plt.plot(
    y_test_forecast.index,
    y_test_forecast.values,
    label="Forecast"
)

plt.xlabel("Date")
plt.ylabel("Number of Trips")
plt.title("Final Test: SARIMA Forecast vs Actual Taxi Trips")

plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

### Final Test Forecast Interpretation

The final SARIMA forecast continued to capture the regular weekly pattern in daily taxi demand.

During October, the forecast generally followed the overall movement of the actual trip counts.

However, larger differences appeared during November and December. The actual number of trips changed more sharply, while the SARIMA forecast remained relatively smooth and continued to follow the regular weekly seasonal pattern.

The largest forecasting errors occurred when actual taxi demand dropped suddenly, particularly near the end of December.

This shows that the model is useful for capturing regular weekly behaviour, but it is less effective at predicting sudden or unusual changes in taxi demand.

```python
# Residual diagnostics for the final SARIMA model
residuals = final_sarima_result.resid

fig = final_sarima_result.plot_diagnostics(
    figsize=(12, 8)
)

plt.tight_layout()
plt.show()
```

### Residual Diagnostics

Residual diagnostics were used to evaluate the final **SARIMA (1,1,0) × (1,0,1,7)** model.

The standardized residuals generally fluctuate around zero without a clear long-term trend.

The histogram is reasonably close to a normal distribution, although there is some deviation in the tails.

The Q-Q plot shows that most residuals are close to the expected straight line, while some observations at the lower and upper ends deviate from the line. This suggests that a small number of relatively large forecasting errors remain.

The correlogram shows that most residual autocorrelations are relatively small, although some correlation remains around the seasonal lags.

Overall, the residual diagnostics suggest that the model captured much of the time series structure, but some unexplained variation may still remain.

```python
from statsmodels.stats.diagnostic import acorr_ljungbox

# Ljung-Box test for residual autocorrelation
ljung_box = acorr_ljungbox(
    final_sarima_result.resid,
    lags=[7, 14],
    return_df=True
)

ljung_box
```

### Ljung-Box Residual Test

The Ljung-Box test was used to check whether significant autocorrelation remained in the residuals of the final SARIMA model **(Statsmodels Developers, 2026)**.

The p-values were:

- **Lag 7:** 0.1645
- **Lag 14:** 0.4032

Both p-values are greater than **0.05**.

Therefore, there is not enough evidence to suggest that significant autocorrelation remains in the residuals.

This indicates that the final **SARIMA (1,1,0) × (1,0,1,7)** model captured most of the important temporal structure in the daily taxi trip data.

### Time Series Conclusion

The daily NYC Yellow Taxi trip data showed clear changes in demand throughout 2025 and a strong weekly seasonal pattern. Saturday had the highest average taxi demand, while Monday had the lowest.

The IQR analysis identified one major anomaly on **25 December 2025**, when the daily number of trips decreased to **65,262**.

The Augmented Dickey-Fuller test was applied to the January to June training data. The original training series had a p-value of **0.0706**, so it was not considered stationary at the 5% significance level.

After applying one level of differencing, the p-value decreased to **0.000097**, indicating that the differenced series was stationary. Therefore, **d = 1** was selected.

The ACF and PACF plots showed strong correlations around lags **7, 14, 21 and 28**, supporting the use of a **7-day seasonal period**.

Several SARIMA models were compared using the July to September validation dataset. **SARIMA (1,1,0) × (1,0,1,7)** achieved the best validation performance among the tested models, with:

- **Validation MAE:** 13,094.29 trips
- **Validation RMSE:** 15,733.34 trips
- **AIC:** 3784.38

The selected model was retrained using January to September data and evaluated on the unseen October to December test dataset.

The final test results were:

- **Final Test MAE:** 17,972.73 trips
- **Final Test RMSE:** 26,354.91 trips

The model captured the regular weekly seasonal pattern reasonably well, but it was less accurate during sudden or unusual changes in taxi demand, especially toward the end of December.

Residual diagnostics and the Ljung-Box test showed no significant remaining autocorrelation at lags 7 and 14, with p-values of **0.1645** and **0.4032**.

Overall, the final SARIMA model captured most of the regular temporal structure in daily taxi demand, although unusual changes remained more difficult to predict.

## Final Conclusion

This project applied unsupervised machine learning and time series forecasting to the 2025 NYC Yellow Taxi dataset.

For clustering, K-Means and Agglomerative Hierarchical Clustering were compared using the same sample of taxi trips. K-Means achieved a higher **Silhouette Score of 0.6953** and a lower **Davies-Bouldin Index of 0.6979**, so it provided the better clustering result.

The final K-Means model identified two main types of taxi trips. Cluster 0 mainly represented shorter and lower-cost trips, while Cluster 1 represented longer and more expensive trips.

For the time series analysis, daily taxi demand showed a strong weekly seasonal pattern. The ADF test showed that one level of differencing was required, so **d = 1** was selected.

Several SARIMA models were compared using the validation dataset, and **SARIMA (1,1,0) × (1,0,1,7)** was selected as the final model.

The final SARIMA model achieved:

- **Test MAE:** 17,972.73 trips
- **Test RMSE:** 26,354.91 trips

The model captured the regular weekly pattern reasonably well, but sudden and unusual changes in taxi demand were more difficult to predict.

Overall, the clustering analysis identified different taxi trip patterns, while the time series analysis provided insight into daily taxi demand and its weekly seasonal behaviour.

## Limitations

This project has several limitations.

First, K-Means clustering was performed using a sample of **100,000 taxi trips** instead of the full dataset because the complete dataset was very large and required significant processing time and memory.

Agglomerative Hierarchical Clustering used a smaller sample of **5,000 trips** because this algorithm requires more computational resources than K-Means.

The clustering analysis used only four numerical features: `trip_distance`, `trip_duration_minutes`, `distance_per_hour`, and `total_amount`. Other features may provide additional information about taxi trip behaviour.

For the time series analysis, only **2025 data** was used. A longer historical period may provide more information about longer-term changes in taxi demand.

The SARIMA model captured the regular weekly pattern but was less accurate when sudden or unusual changes occurred, especially during December.

External factors such as weather, holidays, traffic conditions, and special events were not included in the forecasting model.

## Future Improvements

Several improvements could be made in future work.

For clustering, a larger sample or the full taxi dataset could be used if more computing resources are available.

Additional features such as pickup location, drop-off location, payment type, and time of day could also be included to identify more detailed taxi trip patterns.

Other clustering algorithms, such as DBSCAN, could be tested and compared with K-Means and Agglomerative Hierarchical Clustering.

For time series forecasting, multiple years of historical taxi data could be used to help the model learn longer-term changes and seasonal patterns.

External factors such as weather, public holidays, traffic conditions, and major events could also be included because these factors may influence taxi demand.

Other forecasting models could also be compared with SARIMA to determine whether prediction accuracy can be improved.
