### Time-Based Data Split

The dataset was split based on time to avoid data leakage.

- **January to June:** Training set
- **July to September:** Validation set
- **October to December:** Test set

The training set is used to train the model. The validation set is used to compare models and tune parameters. The test set is used at the end to evaluate the final model on unseen future data.

```python
# Remove trips with zero or negative duration
rows_before = len(df_clean)

df_clean = df_clean[
    df_clean["trip_duration_minutes"] > 0
].copy()

print("Rows removed:", rows_before - len(df_clean))
print("Rows remaining:", len(df_clean))
```

# Part 1: Clustering Analysis

In this section, clustering algorithms will be used to identify different groups of NYC Yellow Taxi trips based on their trip characteristics.

```python
# Select features for clustering
clustering_features = [
    "trip_distance",
    "trip_duration_minutes",
    "distance_per_hour",
    "total_amount"
]

df_clustering = df_analysis[clustering_features].copy()

print("Clustering data shape:", df_clustering.shape)
df_clustering.head()
```

### Clustering Features

Four numerical features were selected for the clustering analysis:

- `trip_distance`
- `trip_duration_minutes`
- `distance_per_hour`
- `total_amount`

These features describe the distance, duration, speed and cost of each taxi trip.

Clustering is an unsupervised learning method, so there is no target variable.

```python
# Sample data for clustering
df_clustering_sample = df_clustering.sample(
    n=100000,
    random_state=42
)

print("Clustering sample shape:", df_clustering_sample.shape)

df_clustering_sample.head()
```

### Clustering Sample

The cleaned dataset contains more than 21 million taxi trips.

A random sample of **100,000 records** was selected for the clustering analysis to reduce processing time and memory usage.

The sample contains the same four clustering features:

- `trip_distance`
- `trip_duration_minutes`
- `distance_per_hour`
- `total_amount`

The sampling used `random_state=42` so that the same records can be selected again when the code is rerun.

```python
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Standardize the clustering sample
clustering_scaled = scaler.fit_transform(df_clustering_sample)


# Convert scaled data back to a DataFrame
df_clustering_scaled = pd.DataFrame(
    clustering_scaled,
    columns=clustering_features
)

print("Scaled data shape:", clustering_scaled.shape)
df_clustering_scaled.head()
```

### Feature Scaling

The four clustering features have different numerical ranges.

`StandardScaler` was used to standardize the features before applying the clustering algorithms.

This gives each feature a similar scale, so variables with larger values do not have too much influence on the clustering results.

After scaling, the dataset contains **100,000 records and 4 standardized features**.

```python
# Define features and target for training data
feature_columns = [
    "PULocationID",
    "DOLocationID",
    "payment_type",
    "trip_distance",
    "trip_duration_minutes",
    "distance_per_hour",
    "pickup_hour",
    "pickup_dayofweek",
    "pickup_month"
]

X_train = df_analysis[feature_columns]
y_train = df_analysis["total_amount"]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
```

### Training Features and Target

The training dataset contains **21,999,037 records**.

The model uses **9 input features** to predict `total_amount`.

`X_train` contains the input features, while `y_train` contains the target variable.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Test different numbers of clusters
k_values = range(2, 7)
inertia_values = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(df_clustering_scaled)
    inertia_values.append(kmeans.inertia_)

print("K values:", list(k_values))
print("Inertia values:", inertia_values)
```

```python
plt.figure(figsize=(7, 5))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for K-Means")

plt.show()
```

### Elbow Method Results

The Elbow Method was used to test K-Means with different numbers of clusters from **2 to 6**.

The inertia decreased as the number of clusters increased.

The largest reduction happened between **k = 2** and **k = 3**, and the rate of improvement became smaller after this point.

Based on the elbow graph, **k = 3** appears to be a reasonable number of clusters.

However, the Silhouette Score and Davies-Bouldin Index will also be used to evaluate the clustering performance before selecting the final value of `k`.

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score

silhouette_scores = []
db_scores = []

for k in range(2, 7):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(df_clustering_scaled)

    # Silhouette Score
    sil_score = silhouette_score(
        df_clustering_scaled,
        labels,
        sample_size=10000,
        random_state=42
    )

    # Davies-Bouldin Index
    db_score = davies_bouldin_score(
        df_clustering_scaled,
        labels
    )

    silhouette_scores.append(sil_score)
    db_scores.append(db_score)

    print(
        f"k={k} | "
        f"Silhouette Score={sil_score:.4f} | "
        f"Davies-Bouldin Index={db_score:.4f}"
    )
```

### K-Means Evaluation Results

The K-Means models were evaluated using the Silhouette Score and Davies-Bouldin Index for different values of `k`.

The results show that **k = 2** achieved the best clustering performance.

- **Silhouette Score:** 0.6898
- **Davies-Bouldin Index:** 0.7095

The Silhouette Score was highest for `k = 2`, while the Davies-Bouldin Index was also the lowest.

Therefore, **2 clusters were selected for the final K-Means model**.

```python
# Train the final K-Means model
final_kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)  

cluster_labels = final_kmeans.fit_predict(df_clustering_scaled)

df_clustering_sample["Cluster"] = cluster_labels

print(df_clustering_sample["Cluster"].value_counts())
```

### Final K-Means Clustering

The final K-Means model was trained using **2 clusters**, based on the previous clustering evaluation results.

The 100,000 sampled taxi trips were divided into:

- **Cluster 0:** 87,750 trips
- **Cluster 1:** 12,250 trips

Cluster 0 contains most of the taxi trips, while Cluster 1 represents a smaller group.

Further analysis is needed to understand the characteristics of each cluster.

```python
# Compare the characteristics of each cluster
cluster_summary = df_clustering_sample.groupby("Cluster")[
    [
        "trip_distance",
        "trip_duration_minutes",
        "distance_per_hour",
        "total_amount"
    ]
].mean()

cluster_summary
```

### Cluster Characteristics

The two K-Means clusters show clear differences in taxi trip characteristics.

**Cluster 0** represents shorter and lower-cost taxi trips.

- Average distance: **2.04 miles**
- Average duration: **13.01 minutes**
- Average speed: **9.64 mph**
- Average total amount: **21.83**

**Cluster 1** represents longer and higher-cost taxi trips.

- Average distance: **12.84 miles**
- Average duration: **38.70 minutes**
- Average speed: **22.10 mph**
- Average total amount: **73.16**

Overall, the K-Means model separated the taxi trips mainly into a group of shorter urban trips and a smaller group of longer, more expensive trips.

```python
# Sample data for cluster visualization
plot_sample = df_clustering_sample.sample(
    n=5000,
    random_state=42
)

plt.figure(figsize=(8, 6))

for cluster in sorted(plot_sample["Cluster"].unique()):
    cluster_data = plot_sample[plot_sample["Cluster"] == cluster]

    plt.scatter(
        cluster_data["trip_distance"],
        cluster_data["total_amount"],
        alpha=0.4,
        label=f"Cluster {cluster}"
    )

plt.xlabel("Trip Distance")
plt.ylabel("Total Amount")
plt.title("K-Means Clusters: Trip Distance vs Total Amount")
plt.legend()

plt.show()
```

```python
# Compare R² scores
model_names = ["Linear Regression", "Decision Tree"]
r2_values = [r2, tree_r2]

plt.figure(figsize=(7, 5))

plt.bar(model_names, r2_values)

plt.ylabel("R² Score")
plt.title("Model R² Comparison")
plt.ylim(0, 1)

plt.show()
```

### R² Score Comparison

The R² scores of the two models were compared.

The higher R² score shows that the Decision Tree explains more of the variation in `total_amount` and performs better on the validation dataset.

### Model Selection

The Decision Tree model was selected as the final model because it performed better than Linear Regression on the validation dataset.

It achieved a lower MAE and RMSE and a higher R² score.

Therefore, the Decision Tree will be used for the final evaluation on the test dataset.
